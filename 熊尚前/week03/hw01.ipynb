```python
# 导包
from torchvision.datasets import KMNIST
from torchvision.transforms import ToTensor
from torch.utils.data import DataLoader
from torch import nn
import torch
```

```python
# 准备数据
# 从指定路径 './data' 加载 KMNIST 测试数据集
# 如果数据不存在，则自动下载
# 使用 ToTensor() 转换将图像数据转换为 PyTorch 张量
train_data = KMNIST(root='./data',train=True,download=True,transform=ToTensor())
test_data = KMNIST(root='./data',train=False,download=True,transform=ToTensor())
# 创建训练数据的数据加载器
# 每次从训练数据集中取出 64 个样本作为一个批次
# shuffle=True 表示每个训练周期开始时打乱数据顺序
train_dl = DataLoader(train_data,batch_size=64,shuffle=True)
test_dl = DataLoader(test_data,batch_size=64,shuffle=False)
print(train_dl)
```

```python
target_dir = {}
for image, target in train_data:
    # print(target)
    if target not in target_dir:
        target_dir[target] = 1
    else:
        target_dir[target] += 1
print(target_dir)
```

```python
# 定义模型
model = nn.Sequential(
    nn.Linear(28*28, 128),
    nn.Sigmoid(),
    nn.Linear(128, 10)
)
```

```python
# 损失函数&优化器
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
for epoch in range(20):
    for data, target in train_dl:
        output = model(data.reshape(-1, 28*28))
        loss = loss_fn(output, target)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    print(f'Epoch:{epoch} Loss:{loss.item()}')    
```

```python
# 测试
total = 0
correct = 0
with torch.no_grad():
    for data, target in test_dl:
        output = model(data.reshape(-1, 28*28))
        _, predicted = torch.max(output, 1)  # 返回每行最大值和索引
        total += target.size(0)
        correct += (predicted == target).sum().item()
print(f'Accuracy: {correct/total*100}%')
```
