```python
from tqdm import tqdm
import torch.nn as nn
import pickle as pkl
```

```python
min_freq = 5
max_size = 10000
# 未知字和 padding 符号
UNK, PAD = '<UNK>', '<PAD>'
vocab_dic = {}
with open('data/train.txt', 'r', encoding='utf-8') as f:
    for line in tqdm(f):
        line = line.strip()
        if not line:
            continue
        content, label = line.split('\t')
        label = int(label)
        # 统计每个词的频率 按照字符分词
        for word in content:
            vocab_dic[word] = vocab_dic.get(word, 0) + 1;
# 筛选出词频大于等于 min_freq 的词，并按词频降序排序，取前 max_size 个
vocab_list = sorted([el for el in vocab_dic.items() if el[1] > min_freq], key=lambda x: x[1], reverse=True)[:max_size]
# 构建词到索引的字典
vocab_dic = {vocab[0]:index for index, vocab in enumerate(vocab_list)}
# 添加未知字和 padding 符号到词表
vocab_dic.update({UNK: len(vocab_dic), PAD: len(vocab_dic) + 1})
pkl.dump(vocab_dic, open('data/vocab.pkl','wb'))

```

```python
with open('data/train.txt','r',encoding='utf-8') as f:
    contents = []
    for line in tqdm(f):
        line = line.strip()
        if not line:
            continue
        content, label = line.split('\t')
        token = [word for word in content]
        seq_len = len(token)
        if seq_len < 32:
            token.extend(PAD * (32 - seq_len))
        else:
            token = token[:seq_len]
        words_line = []
        for word in token:
            words_line.append(vocab_dic.get(word, vocab_dic.get(UNK)))
    contents.append((words_line, int(label), 32))   
```

```python
# 构建模型
class TextCNN(nn.Model):
    def __init__(self):
        self.train_path = 'data/train.txt'
        self.dev_path = 'data/dev.txt'
        self.test_path = 'data/test.txt'
        self.class_list = [x.trip() for x in open('data/class.txt','r', encoding='utf-8').readlines()]
        self.vocab_path = 'data/vocab.pkl'
        # 模型训练结果保存路径
        self.save_path = 'data/save/TextCNN.ckpt'
        self.dropout = 0.5
        # 类别数
        self.num_classes = len(self.class_list)
        # 训练轮数
        self.epochs = 20
        self.learning_rate = 0.001
        # 卷积核尺寸
        self.filter_sizes = (2,3,4)
        # 嵌入层
        self.embedding = nn.Embedding()

        
```
