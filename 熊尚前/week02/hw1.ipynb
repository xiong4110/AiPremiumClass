```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 计算损失函数
def loss(theta):
    return (theta - 2.5) ** 2 - 1

# 计算梯度
def derivate(theta):
    return 2 * (theta - 2.5)
```

```python
# 梯度下降
def gradient_descent(init_theta, eta, epsilon=1e-8):
    # 收集theta的值
    theta_list = [init_theta]

    theta = init_theta
    while True:
        # 计算梯度
        gradient = derivate(theta)
        # 记录上一次的theta值
        last_theta = theta
        # 更新参数
        theta = last_theta - eta * gradient
        print(theta)
        theta_list.append(theta)
        # 计算损失函数
        loss_v = loss(theta)
        if abs(loss_v - loss(last_theta)) < epsilon:
            break
    return theta_list
```

```python
plot_x = np.linspace(-1, 6, 150)
plot_y = (plot_x - 2.5) ** 2 - 1
```

```python
def plot_theta(theta_list):
    plt.plot(plot_x,plot_y)
    plt.plot(theta_list, loss(np.array(theta_list)))
    # 打印最终的theta值
plot_theta(gradient_descent(0, 0.1))
```
