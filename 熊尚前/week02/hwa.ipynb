### 使用sklearn数据集（鸢尾花）训练逻辑回归模型

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
# 生成训练数据,加载鸢尾花数据集
X,y = load_iris(return_X_y=True)
print(y)

# 由y的结果可知，鸢尾花有三种类型，每种类型50个样本，我们只取前100个，这样就变成了只有两种类型的数据集，就变成了二分类问题
X = X[:100]
y = y[:100]
print(y)
# 拆分训练集、测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=100)

```

```python
# 权重参数
theta = np.random.randn(1, 4) # 有4个特征
bias = 0

# 超参数
lr = 0.01
epochs = 500
```

```python
# 模型函数
def forward(x, theta, bias):
    z = np.dot(theta, x.T) + bias # 线性运算
    return (1/(1+np.exp(-z))) # 用sigmoid函数将线性运算结果映射成概率值

# 损失函数
def loss_func(y, y_hat):
    e = 1e-8
    return -y*np.log(y_hat + e) - (1-y)*np.log(1-y_hat + e) # 二元交叉熵损失函数

# 计算梯度
def calc_gradient(x,y,y_hat):
    m = x.shape[0] # 样本数量
    dw = np.dot((y_hat - y), x)/m # 权重梯度
    db = np.mean(y_hat - y) # 偏置项梯度
    return dw, db
```

```python
# 模型训练
for i in range(epochs):
    # 前向传播
    y_hat = forward(X_train, theta, bias)
    # 计算损失
    loss = loss_func(y_train, y_hat)
    # 计算梯度
    dw, db = calc_gradient(X_train, y_train, y_hat)
    # 更新参数
    theta = theta - lr * dw
    bias = bias - lr * db

    # 打印损失值
    if i % 100 == 0:
        # 计算准确率
        acc = np.mean(np.round(y_hat) == y_train)
        print(f"epoch {i}, loss: {np.mean(loss)}, acc: {acc}")
```

```python
# 模型推理
for i in range(len(y_test)):
    y_hat = forward(X_test[i], theta, bias)
    print(f"y_hat: {y_hat}, round: {np.round(y_hat)}, y: {y_test[i]}")
```
