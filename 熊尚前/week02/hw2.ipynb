```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import numpy as np
```

```python

X,y = make_classification()
# 将数据拆分，拆成训练集和测试集
train_X,test_X,train_y,test_y=train_test_split(X,y,test_size=0.2)
```

```python
# 准备参数
theta = np.random.randn(1,20)
bias = 0
lr = 0.001
epoch = 5000
```

```python
# 模型运算
def forward(x,theta,bias):
    z = np.dot(theta, x.T) + bias
    return (1/(1+np.exp(-z)))
```

```python
# 计算损失
def loss_function(y, y_hat):
    e = 1e-8
    return - y * np.log(y_hat + e) - (1-y) * np.log(1-y_hat+e)
```

```python
# 计算梯度
def calc_gradient(x,y,y_hat):
    m = x.shape[-1]
    delta_w = np.dot(y_hat-y,x)/m
    delta_b = np.mean(y_hat-y)
    return delta_w, delta_b
```

```python
# 重复，观察损失函数，调整学习率
for i in range(epoch):
    # 前向传播
    y_hat = forward(train_X, theta, bias)
    # 计算损失
    loss = np.mean(loss_function(train_y, y_hat))
    if i % 100 == 0:
        print('step:{}, loss:{}'.format(i, loss))
    # 梯度下降
    dw, db = calc_gradient(train_X, train_y, y_hat)
    # 更新参数
    theta -= lr * dw
    bias -= lr * db
```

```python
# 测试模型
idx = np.random.randint(len(test_X))
x = test_X[idx]
y = test_y[idx]

def predict(x):
    y_hat = forward(x, theta, bias)
    return 1 if y_hat > 0.5 else 0
pred = predict(x)
print('预测值：', pred, '真实值：', y)
```

```python
# pytorch实现逻辑回归
import torch
from sklearn.datasets import load_iris
from sklearn.datasets import make_classification

# 学习率，控制参数更新的步长
lr = 0.01
# 数据准备, 生成1000个样本，每个样本有10个特征的分类数据集
X, y = make_classification(n_samples=1000, n_features=10)

# 将numpy数组转换为PyTorch张量，方便后续在PyTorch中进行计算
tensor_x = torch.tensor(X, dtype=torch.float)
tensor_y = torch.tensor(y, dtype=torch.float)

# 创建模型参数并初始化，w为权重矩阵，b为偏置项
# requires_grad=True 表示需要对这些参数进行梯度计算
w = torch.randn(1, 10, requires_grad=True)
b = torch.randn(1, requires_grad=True)

# 训练循环，迭代5000次
for i in range(5000):
    # 前向运算
    # 计算线性组合，公式为：r = X * w^T + b
    r = torch.nn.functional.linear(tensor_x, w, b)
    # 通过sigmoid函数将线性输出转换为概率值
    r = torch.sigmoid(r)

    # 计算损失，使用二元交叉熵损失函数
    # reduce='mean' 表示对损失求平均值
    loss = torch.nn.functional.binary_cross_entropy(r.squeeze(), tensor_y, reduction='mean')
    
    # 反向传播，计算损失函数关于参数w和b的梯度
    loss.backward()

    # 关闭自动求导，手动更新参数
    with torch.autograd.no_grad():
        # 更新权重参数，使用梯度下降法：w = w - lr * grad(w)
        w -= lr * w.grad
        # 梯度清零，避免下次计算时梯度累积
        w.grad.zero_()
        # 更新偏置参数，使用梯度下降法：b = b - lr * grad(b)
        b -= lr * b.grad
        # 梯度清零，避免下次计算时梯度累积
        b.grad.zero_()
    # 打印当前训练轮次的损失值，保留四位小数
    print(f'train loss:{loss.item():.4f}')
```
